```python
# Dirty Data Handling: You have a list ['10', '20', 'abc', '40']. Use a loop to convert them to integers, 
# skipping the ones that cause a ValueError.

l = ['10','20','30','abc','40']
for i in l:
    try:
        num = int(i)
        print(num)      
    except ValueError:
        print(f'can not convert {i} into the integer')
```

```python
# Manual Raising: Write a function check_balance(amount). If the amount is negative, raise ValueError("Balance cannot be negative").

def check_balance(amount):
    try:
        if amount < 0:
            print('balance cannot be nagative')
        else:
            print('yes,you can withdraw money')
    except Exception as e:
        print(e)
```

```python
# Custom Exceptions: Create your own exception class InsufficientFundsError(Exception) and raise it in a simulated bank withdrawal.
class InsufficientFundsError(Exception):
    pass


def bank(balance,amount):
    if amount > balance:
            raise InsufficientFundsError('Insufficient Funds Error')
    else:
            print('yes you can withdraw money')
try:
    bank(15000,7000)
except InsufficientFundsError as e:
        print(e)
```

```python
# Nested Try-Except: Put a try-except block inside another except block to try a backup solution if the first error-handling fails.
try:
    f = open('fi1.txt',mode = 'r')  # if we write the w+ so this not give any error but create new file
except:
    try:
        f = open('fi2.txt',mode = 'r')
    except:
        print('Not found')
```

```python
# Logging Errors: Instead of just printing an error, use the logging module to save the error message to a file
# called error.log.

logging.basicConfig(filename = 'error.log',level = logging.INFO,
                   format = '%(asctime)s - %(levelname)s - %(message)s' )
```

```python
def division(a,b):
    logging.info('Program Start here.')
    try:
        div = a / b
        logging.info(f'You enter {a} and {b} as input.')
        logging.info(f'the result is {div}.')
        logging.info('------------------------------')
    except Exception as e:
        logging.error(e)
        logging.error ('Invalid input')
        logging.error('-----------------------------')
```

```python
division(20,0)
```

```python
# Retrying Logic: (Conceptual) Use a while loop with a try-except block to give a user 3 chances to enter a
# valid password before locking them out.

count = 0
while count < 3:
    a = input('enter the password: ')
    count = count + 1
    try:
        if len(a) == 8:
            print('valid pasword')
            break
        else:
            print('invalid password')
    except Exception as e:
        print(e)
else:
    print('3 chance you have and now they are over')
    print('try next time')
```
