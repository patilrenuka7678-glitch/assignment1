```python
#  Q.1  Create a class Laptop with a constructor that initializes brand, ram, and processor.
# Create an object of the Laptop class called my_laptop with values ("Dell", "16GB", "i7").
# Add a method (function) inside the Laptop class called show_specs() that prints all the laptop's details.

class laptop():
    def __init__(self,brand,ram,processor):
        self.brand = brand
        self.ram = ram
        self.processor = processor
    def show_specs(self):
        print(self.brand)
        print(self.processor)
        print(self.ram)
    def __str__(self):
        return f"{self.brand} Laptop - {self.processor} Processor - {self.ram} RAM"
```

```python
my_laptop = laptop('Dell','16GB','i7')
```

```python
my_laptop.show_specs()
```

```python
print(my_laptop)
```

```python
# Area Calculator: Create a class Circle that takes radius. Add a method calculate_area() (use 3.14 * r * r).

class circle():
    def __init__(self, r):
        self.r = r
    def calculate_area(self):
        self.area = 3.14 * self.r * self.r
        return self.area
```

```python
c = circle(20)
```

```python
c.calculate_area()
```

```python
# Status Checker: In a SmartDevice class, add a variable is_on = False. Create a method toggle() that switches the state.
# ---> here is toggle() so if value is true so it's return flase when flase so true

class smartdevice():
    def __init__(self,is_on):
        self.is_on = is_on
    def toggle(self):
        self.is_on = not self.is_on
        return self.is_on
```

```python
sd = smartdevice(True)
```

```python
sd.toggle()
```

```python
# Object Counting: Create a class where a variable count increases every time a new object is created.

class obj_count():
    count = 0     # class variable
    def __init__(self):
        obj_count.count = obj_count.count + 1    # we can access the class variable by using the 'classname.variablename'
```

```python
# List of Objects: Create a loop that generates 5 car objects with different names and stores them in a list.

l = []
l1 = ['audi','BMW','Vagonar','TATA','Mahindra']
class car():
    def __init__(self,name):
        self.name = name
for i in l1:
        obj = car(i)
        l.append(obj)
for car_obj in l:
        print(car_obj.name)
```
