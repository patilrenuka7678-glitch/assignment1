```python
# Pass/Fail: Given marks = [20, 45, 80, 30], create a list that says "Pass" if marks ≥ 35, else "Fail".

marks = [20,45,80,30]
l = []

for i in marks:
    if i >= 35:
        l.append('pass')
    else:
        l.append('fail')
print(l)
```

```python
# By Using the comprehension......
marks = [20,45,80,30]
con = ['pass' if i >= 35 else 'fail' for i in marks]
```

```python
# List Flattening with Filter: Flatten [[1, 2], [3, 4]] but only if the sub-list sum is greater than 3.

l = [[1,2],[3,4]]
s = []

for i in l:
    if sum(i) > 3:
        for j in i:
            s.append(j)
print(s)
```

```python
l = [[1,2],[3,4]]
con16 = [j for i in l if sum(i) > 3 for j in i]  # nested for loop
print(con16)
```

```python
# Conditional Dictionary Swap: Swap keys and values in a dictionary, but only if the value is a string.

dict_swap = {1:'hello',2:200,3:'python',4:16}
con17 = {v if type(v) == str else k:   # here we apply the two conditions 
                                       # 1st that check the value if value is true so print the value then print key when second 
                                       # condition becomes true and when condition becomes false it print key value pair as it is
         k if type(v) == str else v
         for k,v in dict_swap.items()}  # .items() is used to get key:values one by one [Note: It work like loop]
```
